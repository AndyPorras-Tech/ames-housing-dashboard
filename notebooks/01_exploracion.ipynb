```python
print("Copia para Machine Learning, revisar nulos, tratar nulos, convertir a numeros")
```

```python
df_ml = df_clean.copy()

df_ml.isnull().sum().sort_values(ascending=False).head(20)
```

```python
cat_cols = [
    "Mas Vnr Type",
    "Fireplace Qu",
    "Garage Cond",
    "Garage Finish",
    "Garage Qual",
    "Garage Type",
    "Bsmt Exposure",
    "BsmtFin Type 2",
    "BsmtFin Type 1",
    "Bsmt Cond",
    "Bsmt Qual"
]

for col in cat_cols:
    df_ml[col] = df_ml[col].fillna("None")
```

```python
num_cols = [
    "Lot Frontage",
    "Garage Yr Blt",
    "Mas Vnr Area",
    "Bsmt Half Bath",
    "Bsmt Full Bath",
    "TotalBathrooms",
    "Garage Cars",
    "BsmtFin SF 1",
    "Bsmt Unf SF"
]

for col in num_cols:
    df_ml[col] = df_ml[col].fillna(df_ml[col].median())
```

```python
df_ml.isnull().sum().sum()
```

```python
df_ml.isnull().sum().sort_values(ascending=False)
```

```python
df_ml["Electrical"] = df_ml["Electrical"].fillna(
    df_ml["Electrical"].mode()[0]
)

df_ml["Total Bsmt SF"] = df_ml["Total Bsmt SF"].fillna(
    df_ml["Total Bsmt SF"].median()
)

df_ml["Garage Area"] = df_ml["Garage Area"].fillna(
    df_ml["Garage Area"].median()
)

df_ml["BsmtFin SF 2"] = df_ml["BsmtFin SF 2"].fillna(
    df_ml["BsmtFin SF 2"].median()
)
```

```python
df_ml.isnull().sum().sum()
```

```python
df_ml = pd.get_dummies(
    df_ml,
    drop_first=True
)
```

```python
df_ml.shape
```

```python
df_ml.info()
```

```python
print("Aqui empieza el modelado xD")
```

```python
x = df_ml.drop("SalePrice", axis=1)

y = df_ml["SalePrice"]
```

```python
x.shape
```

```python
y.shape
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(

x,
y,
test_size=0.2,
random_state=42

)
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
print("Siguiente paso: Importar el modelo")
```

```python
from sklearn.linear_model import LinearRegression
```

```python
print("Ahora crear el modelo")
```

```python
model = LinearRegression()
```

```python
print("Siguiente paso: entrenar modelo")
```

```python
model.fit(x_train, y_train)
```

```python
y_pred = model.predict(x_test)
```

```python
print("Evaluar modelo")

from sklearn.metrics import(
mean_absolute_error,
mean_squared_error,
r2_score

)

import numpy as np
```

```python
mae = mean_absolute_error(y_test, y_pred)

mse = mean_squared_error(y_test, y_pred)

rmse = np.sqrt(mse)

r2 = r2_score(y_test, y_pred)


print("MAE:", mae )
print("MSE:", mse )
print("RMSE:", rmse )
print("R2:", r2 )
```

```python
print("Comparacion")

comparacion = pd.DataFrame({

    "Real": y_test,
    "Predicho": y_pred
})

comparacion.head(10)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,6))

plt.scatter(y_test, y_pred)

plt.xlabel("Precio real")
plt.ylabel("Precio Predicho")

plt.title("Real vs Predicho")

plt.show()
```

```python
print("Modelo Decision Tree Regressor")


from sklearn.tree import DecisionTreeRegressor
```

```python
print("Evaluacion")

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

import numpy as np

mae_tree = mean_absolute_error(y_test, y_pred_tree)

mse_tree = mean_squared_error(y_test, y_pred_tree)

rmse_tree = np.sqrt(mse_tree)

r2_tree = r2_score(y_test, y_pred_tree)

print("MAE :", mae_tree)
print("MSE :", mse_tree)
print("RMSE :", rmse_tree)
print("R2 :", r2_tree)

```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,6))

plt.scatter(y_test, y_pred_tree)

plt.xlabel("Precio Real")
plt.ylabel("Precio Predicho")

plt.title("Decision Tree - Real vs Predicho")

plt.show()
```
